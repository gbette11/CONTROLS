```python
import numpy as mp
import matplotlib.pyplot as plt
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
```

```python
def main():
    #2 List
    list_a = [3, 4, 6, 10, 39, 87, 12, 54]
    print("this is the max:", max(list_a))
    print("this is the min:", min(list_a))
    print("this is the sum:", sum(list_a))
    print("sorted:", sorted(list_a))
    for x in list_a:
        print(x)

    #3 Prime
    def prime(n):
        if n < 2:
            return False
        for i in range(2, int(n**0.5)+1):
            if n % i == 0:
                return False
        return True
    print ("prime # below 100:", [n for n in range(100) if prime(n)])
    def ncustom(n):
        return [num for num in range(n) if prime(num)]
    print("custom prime #:", ncustom(1000))
    
    #4
    def create_3d_array(n, data):
    
        if data == 'zeros':
            return mp.zeros((n, 3, 3))  # Create an array filled with zeros
        elif data == 'random':
            return mp.random.randint(1, 101, size=(n, 3, 3))  # Create an array with random integers between 1 and 100
        else:
            raise ValueError("Invalid 'data' argument. Use 'zeros' or 'random'.")
    array_zeros = create_3d_array(10, 'zeros')
    print("Array with zeroes (n=10):")
    print(array_zeros)    
    array_random = create_3d_array(5, 'random')
    print("\nArray with random numbers between 1 and 100 (n=5):")
    print(array_random)
    
    #5
    A = mp.array([[3, 4, -5],
            [0, 5, 1],
            [1, 3, 6]])
    B = mp.array([10, 8, -5])
    solution = mp.linalg.solve(A, B)
    a, b, c = solution
    print(f"The solution is: a = {a}, b = {b}, c = {c}")
    
    #6
    def f(x):
        return 0.1 * x**3 - x**2 + 5
    x = mp.linspace(-5, 5, 400)  # 400 points from -5 to 5
    y = f(x)
    plt.plot(x, y, label="y = 0.1x^3 - x^2 + 5", color='blue')
    result_min = minimize_scalar(f, bounds=(-5, 5), method='bounded')
    result_max = minimize_scalar(lambda x: -f(x), bounds=(-5, 5), method='bounded')
    min_x = result_min.x
    min_y = f(min_x)
    max_x = result_max.x
    max_y = f(max_x)
    plt.scatter(min_x, min_y, color='red', zorder=5)
    plt.scatter(max_x, max_y, color='green', zorder=5)
    plt.text(min_x, min_y, f'Min ({min_x:.2f}, {min_y:.2f})', color='red', fontsize=12, verticalalignment='bottom')
    plt.text(max_x, max_y, f'Max ({max_x:.2f}, {max_y:.2f})', color='green', fontsize=12, verticalalignment='top')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Plot of y = 0.1x^3 - x^2 + 5')
    plt.axhline(0, color='black',linewidth=0.5)
    plt.axvline(0, color='black',linewidth=0.5)
    plt.grid(True)
    plt.legend()
    plt.show()
```

```python
if __name__ == "__main__":
    main()
```
